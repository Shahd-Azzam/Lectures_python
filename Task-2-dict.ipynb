```python
person = {
    'first_name': 'Milaan',
    'last_name': 'Parmar',
    'age': 96,
    'country': 'Finland',
    'is_marred': True,
    'skills': ['Python', 'Matlab', 'R', 'C', 'C++'],
    'address': {
        'street': 'Space street',
        'zipcode': '02210'
    }
}
```

```python
# Q1:

if 'skills' in person:
    print(person['skills'][len(person['skills']) // 2])
else:
    print("skills key does not exist")
```

```python
# Q2:

if 'skills' in person:
    if 'Python' in person['skills']:
        print(True)
    else:
        print(False)
else:
    print("skills key does not exist")
```

```python
# Q3:

if 'skills' in person:
    skills = person['skills']

    if skills == ['Python', 'Matlab']:
        print("He knows machine learning")

    elif 'Python' in skills and 'R' in skills:
        print("He knows statistics")

    elif 'C' in skills and 'C++' in skills:
        print("He knows software development")

    else:
        print("unknown title")

else:
    print("skills key does not exist")
```

```python
# Q4:

if person['is_marred'] and person['country'] == 'Finland':
    print(f"{person['first_name']} {person['last_name']} lives in Finland. He is married.")
else:
    print("The condition is not met")
```
