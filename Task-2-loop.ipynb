```python
from countries import countriesWWWWW
from countries_details_dat import countries_data
```

```python
# Q1:

land_countries = []

for country in countriesWWWWW:
    if "land" in country.lower():
        land_countries.append(country)
    else:
        pass

print(land_countries)
```

```python
# Q2:

languages = set()

for country in countries_data:
    if "languages" in country:
        for language in country["languages"]:
            languages.add(language)
    else:
        pass

print(len(languages))
```

```python
# Q3:

from collections import Counter

languages = Counter()

for country in countries_data:
    if "languages" in country:
        for language in country["languages"]:
            languages[language] += 1
    else:
        pass

print(languages.most_common(10))
```

```python
# Q4:

countries_population = []

for country in countries_data:
    if "population" in country:
        countries_population.append(
            (country["population"], country["name"])
        )
    else:
        pass

countries_population.sort(reverse=True)

print(countries_population[:10])
```
