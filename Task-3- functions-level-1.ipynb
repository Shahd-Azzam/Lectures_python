```python
# Q1:

def area_of_circle(r):
    area = 3.14 * r * r
    return area


def perimeter_of_circle(r):
    perimeter = 2 * 3.14 * r
    return perimeter


r = float(input("Enter radius: "))

print("Area =", area_of_circle(r))
print("Perimeter =", perimeter_of_circle(r))
```

```python
# Q2:

def add_all_nums(*args):
    total = 0

    for num in args:
        if not isinstance(num, (int, float)):
            return "All arguments must be numbers."

        total += num

    return total


numbers = input("Enter numbers separated by spaces: ").split()
numbers = [float(num) for num in numbers]

print(add_all_nums(*numbers))
```

```python
# Q3:

def convert_celsius_2_fahrenheit(celsius):
    fahrenheit = (celsius * 9 / 5) + 32
    return fahrenheit


celsius = float(input("Enter temperature in Celsius: "))

print("Temperature in Fahrenheit =", convert_celsius_2_fahrenheit(celsius))
```

```python
# Q4:

def check_season(month):
    month = month.lower()

    if month in ["september", "october", "november"]:
        return "Autumn"
    elif month in ["december", "january", "february"]:
        return "Winter"
    elif month in ["march", "april", "may"]:
        return "Spring"
    elif month in ["june", "july", "august"]:
        return "Summer"
    else:
        return "Invalid month"


month = input("Enter month: ")

print(check_season(month))
```

```python
# Q5:

def calculate_slope(x1, y1, x2, y2):
    slope = (y2 - y1) / (x2 - x1)
    return slope


x1 = float(input("Enter x1: "))
y1 = float(input("Enter y1: "))
x2 = float(input("Enter x2: "))
y2 = float(input("Enter y2: "))

print("Slope =", calculate_slope(x1, y1, x2, y2))
```

```python
# Q6:

def solve_quadratic_eqn(a, b, c):
    discriminant = b ** 2 - 4 * a * c

    if discriminant > 0:
        x1 = (-b + discriminant ** 0.5) / (2 * a)
        x2 = (-b - discriminant ** 0.5) / (2 * a)
        return x1, x2

    elif discriminant == 0:
        x = -b / (2 * a)
        return x

    else:
        return "No real solutions"


a = float(input("Enter a: "))
b = float(input("Enter b: "))
c = float(input("Enter c: "))

print("Solution =", solve_quadratic_eqn(a, b, c))
```

```python
# Q8:

def capitalize_list_items(items):
    capitalized_list = []

    for item in items:
        capitalized_list.append(item.capitalize())

    return capitalized_list


items = input("Enter list items separated by spaces: ").split()

print(capitalize_list_items(items))
```

```python
# Q11:

def sum_of_numbers(n):
    total = 0

    for i in range(1, n + 1):
        total += i

    return total


n = int(input("Enter a number: "))

print(sum_of_numbers(n))
```

```python
# Q12:

def sum_of_odds(n):
    total = 0

    for i in range(1, n + 1):
        if i % 2 != 0:
            total += i

    return total


n = int(input("Enter a number: "))

print(sum_of_odds(n))
```

```python
# Q13:

def sum_of_even(n):
    total = 0

    for i in range(1, n + 1):
        if i % 2 == 0:
            total += i

    return total


n = int(input("Enter a number: "))

print(sum_of_even(n))
```
