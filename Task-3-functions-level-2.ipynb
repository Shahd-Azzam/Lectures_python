```python
# Q1:

def capitalize_list_items(items):
    capitalized_list = []

    for item in items:
        capitalized_list.append(item.capitalize())

    return capitalized_list


items = input("Enter list items separated by spaces: ").split()

print(capitalize_list_items(items))
```

```python
# Q2:

def add_item(items, item):
    items.append(item)
    return items

while True:
    print("\nEnter your list")
    items = items = input("Enter items separated by spaces: ").split()

    while True:
        item = input("Enter item to add: ")

        add_item(items, item)

        choice = input("Do you want to add another item? (yes/no): ").lower()

        if choice == "no":
            break

    print("Your list:", items)

    choice = input("Do you want to create a new list? (yes/no): ").lower()

    if choice == "no":
        break
```

```python
# Q3:

def reverse_list(items):
    reversed_list = []

    for i in range(len(items) - 1, -1, -1):
        reversed_list.append(items[i])

    return reversed_list


items = input("Enter list items separated by spaces: ").split()

print(reverse_list(items))
```

```python
# Q4:

def evens_and_odds(n):
    evens = 0
    odds = 0

    for i in range(n + 1):
        if i % 2 == 0:
            evens += 1
        else:
            odds += 1

    print("The number of odds are", odds)
    print("The number of evens are", evens)


n = int(input("Enter a positive integer: "))

evens_and_odds(n)
```

```python
# Q5:

def calculate_statistics(numbers):
    # Mean
    mean = sum(numbers) / len(numbers)

    # Median
    numbers_sorted = sorted(numbers)
    n = len(numbers_sorted)

    if n % 2 == 1:
        median = numbers_sorted[n // 2]
    else:
        middle1 = numbers_sorted[n // 2 - 1]
        middle2 = numbers_sorted[n // 2]
        median = (middle1 + middle2) / 2

    # Mode
    counts = {}

    for number in numbers:
        if number in counts:
            counts[number] += 1
        else:
            counts[number] = 1

    mode = numbers[0]

    for number in counts:
        if counts[number] > counts[mode]:
            mode = number


    return mean, median, mode


numbers = input("Enter numbers separated by spaces: ").split()
numbers = [float(x) for x in numbers]

mean, median, mode = calculate_statistics(numbers)

print("Mean =", mean)
print("Median =", median)
print("Mode =", mode)
```
