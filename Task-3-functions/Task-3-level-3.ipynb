```python
# Q1:

def is_prime(number):
    if number < 2:
        return False

    for i in range(2, number):
        if number % i == 0:
            return False

    return True


number = int(input("Enter a number: "))

print(is_prime(number))
```

```python
# Q2:

def are_all_unique(items):
    for i in range(len(items)):
        for j in range(i + 1, len(items)):
            if items[i] == items[j]:
                return False

    return True


items = input("Enter list items separated by spaces: ").split()

print(are_all_unique(items))
```

```python
# Q3:

def are_same_type(items):
    first_type = type(items[0])

    for item in items:
        if type(item) != first_type:
            return False

    return True


items = []

while True:
    value = input("Enter an item: ")

    if value.isdigit():
        value = int(value)
    elif value.replace(".", "", 1).isdigit():
        value = float(value)
    else:
        value = str(value)

    items.append(value)

    choice = input("Do you want to add another item? (yes/no): ").lower()

    if choice == "no":
        break

print(are_same_type(items))
```

```python
# Q4:

def is_valid_variable(variable):
    import keyword

    return variable.isidentifier() and not keyword.iskeyword(variable)


variable = input("Enter a variable name: ")

print(is_valid_variable(variable))
```
