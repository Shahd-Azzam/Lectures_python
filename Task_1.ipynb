```python
# Q1:
my_list = [1, 2, 3, 4, 5]
print(my_list)

my_list.append(6)
print(my_list)

my_list[1] = 10

print(my_list)
```

```python
# Q2:
my_list = ['apple', 'banana', 'cherry', 'date']
print(my_list)

print(my_list[0])

my_list.pop()

print(my_list)

my_list.insert(2, 'grape')

print(my_list)

```

```python
# Q3:
my_set = {1, 2, 3, 4}
my_set.add(5)
print(my_set)

my_set.add(3)
my_set.remove(2)

print(my_set)
```

```python
# Q4:
my_set = {'red', 'green', 'blue'}

print('green' in my_set)

my_set.add('yellow')

my_set.remove('red')

print(my_set)
```

```python
# Q5:
my_tuple = (1, 2, 3, 4, 5)

print(my_tuple[2])

my_tuple[1] = 10 
# This happens because tuples are immutable, meaning their elements cannot be changed after creation.

print(my_tuple) 
```

```python
# Q6:
my_tuple = ('apple', 'banana', 'cherry')

print(my_tuple[-1])

print('banana' in my_tuple)

print(my_tuple)
```

```python
# Q7:
my_dict = {
    'name': 'John',
    'age': 25,
    'city': 'New York'
}

print(my_dict['name'])

my_dict['age'] = 26

my_dict['job'] = 'Engineer'

print(my_dict)
```

```python
# Q8:
my_dict = {
    'brand': 'Toyota',
    'model': 'Corolla',
    'year': 2019
}

print(my_dict['model'])

del my_dict['year']

my_dict['color'] = 'blue'

print(my_dict)
```
